Nominal One-Hot Encoding (OHE) is a data preprocessing technique that converts unordered categorical variables (such as colors, cities, or animal types) into a binary (0 or 1) numerical format for machine learning models.

```python
import pandas as pd 
from sklearn.preprocessing import OneHotEncoder
```

```python
##Create a simple dataframe
df=pd.DataFrame({
    'color':['red','blue','green','red','blue']
})
```

```python
df.head()
```

A One-Hot Encoder is a tool or algorithm that implements the one-hot encoding technique. It converts categorical text or integer data into a format that machine learning models can easily interpret.
Instead of changing the data itself, a One-Hot Encoder acts as the processor that automates the creation of the binary (0 and 1) columns.

Scikit-Learn 

```python
##Create an instance of One Hot Encoder
encoder= OneHotEncoder()
```

```python
#Learn to convert the color into an array why [[]] because OneHotEncoder takes in 2 dimensional
encoded=encoder.fit_transform(df[['color']]).toarray()
```

```python
encoded
```

```python
#get_feature_names_out() retrieve the features names generated by the fitted encoder
encoder_df=pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
encoder_df
```

**Label Encoding**

Label Encoding is a data preprocessing technique that converts categorical values into numerical integers. It assigns a unique integer to each category based on alphabetical order or the order of appearance.
For example, a Size column with values [Small, Medium, Large] would transform into a single numerical column with values like [2, 1, 0].

Disadvantages:- artificial mathematical order (ranking) to unordered data, which can confuse and degrade the performance of many machine learning algorithms.

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
lbl_encoder=LabelEncoder()
df
```

```python
lbl_encoder.fit_transform(df[['color']])
```

```python
#why array(2) comes because of alphabetical order
lbl_encoder.transform([['red']])
```

**Ordinal Encoding**

Ordinal Encoding is a data preprocessing technique that converts categorical values into numerical integers while preserving their natural order or hierarchy.
It is used exclusively for ordinal variables—features where the categories have a clear, meaningful rank (for example: [Low, Medium, High] or [Freshman, Sophomore, Junior, Senior]).

```python
##ordinal Encoding
from sklearn.preprocessing import OrdinalEncoder
```

```python
#Create a simple datafrme with an ordinal
df=pd.DataFrame({
    'size':['small','medium','large','medium','large','small']

})
df
```

```python
#Create an instance of ordinal encoder and then fit_tranform
encoder= OrdinalEncoder(categories=[['small','medium','large']])
encoder.fit_transform(df[['size']])
```

**Target Guided Ordinal Encoding**

Target Guided Ordinal Encoding is a clever trick to convert text categories into numbers based on how much they impact the final answer (the target variable) you are trying to predict.

```python
#Create a simple dataframe with the categorical value and target variable
df=pd.DataFrame({
    'city':['New York','London','Paris','Tokyo','Paris'],
    'price':[200,150,300,234,567]
})
df
```

```python
#to_dict to get into dictionary format
mean_price=df.groupby('city')['price'].mean().to_dict()
mean_price
```

```python
df['city_encoded']=df['city'].map(mean_price)
df
```
