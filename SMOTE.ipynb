### SMOTE(Synthetic Minority Oversampling Technique)

SMOTE (Synthetix Minority Over-Sampling Technique) is a technique used in machine learning to address imbalanced datasets where the minority class has significantly fewer instance than the majority class. SMOTE involves generating synthetic instances of the minority class by interpolating between existing instances.

```python
from sklearn.datasets import make_classification
```

```python
X,y = make_classification(n_samples=1000, n_redundant=0, n_features=2, n_clusters_per_class=1, weights=[0.90], random_state=12) 
# X is my independent feature, y is my dependent feature(target/output). I have 1000 samples, 2 features, 90% of the data belongs to one class and 10% belongs to another class.
```

_redundant=0 means we don't want make_classification() to generate any additional features that are redundant/derived from the other features.

```python
X,y
```

```python
import pandas as pd
df1 = pd.DataFrame(X, columns=['f1','f2'])
df2 = pd.DataFrame(y, columns=['Target'])
final_df = pd.concat([df1,df2], axis=1)
final_df.head()
```

```python
final_df['Target'].value_counts()
```

```python
import matplotlib.pyplot as plt
plt.scatter(final_df['f1'], final_df['f2'], c=final_df['Target'])
```

```python
from imblearn.over_sampling import SMOTE
```

```python
# transform the dataset
oversample=SMOTE()
X,y= oversample.fit_resample(final_df[['f1','f2']], final_df['Target'])
```
