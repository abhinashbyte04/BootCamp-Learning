```python
def is_strong_password(password):
    if len(password)<8:
        return "week password"

    if not any(char.isdigit() for char in password):
        return "week password"
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper()for char in password):
        return False
    if not any(char in '!@#$%^&*()_+' for char in password):
        return False
    return True
print(is_strong_password("weakPwd"))
```

```python
def count_text_file(file_path):

    with open(file_path, 'r') as file:
        lines = file.readlines()

        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)

        return line_count, word_count, char_count


file_path = 'example.txt'

lines, words, characters = count_text_file(file_path)

print(f"Lines: {lines}, Words: {words}, Characters: {characters}")
```

```python
## Merging and joining DataFrames
#create sample DataFrames

df1 = a.DataFrame({'Key': ['A','B','C'],'value1':[1,2,3]})
df2 = a.DataFrame({'Key': ['A','B','D'],'value2':[4,5,6]})

df1
```

```python
df2
```

```python
##Merging DataFrame on the 'Key Column'

a.merge(df1,df2, on="Key",how="inner")
```

```python
a.merge(df1,df2, on="Key",how="outer")
```

```python
a.merge(df1,df2, on="Key",how="left")
```

```python
a.merge(df1,df2, on="Key",how="right")
```
